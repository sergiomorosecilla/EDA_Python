3. Uno los dos datasets a través de sus respectivas columnas 'id' que es la referencia de las clientes y puede servir como clave única. Acabaré de realizar limpieza final de datos

```python
# Realizar outer join (todos los registros de ambos DataFrames)
df_merged = df_bank.merge(df_customer, on='id', how='outer')
# Tamaños
print("Tamaño df_bank:", df_bank.shape)
print("Tamaño df_customer:", df_customer.shape)
print("Tamaño df_merged (outer):", df_merged.shape)

df_merged.info()
df_merged.head()
```

Realizo últimas operaciones de limpieza o mejora de visualización de datos

```python
#Mejoro visibilidad de variables buleanas 'default','housing' y 'loan' para que muestren
#un valor booleano limpio sin decimales y en datos vacíos o NaN muestre un 0.
cols = ['default', 'housing', 'loan']
# Reemplazar valores vacíos o NaN con 0 y convertir todos los valores a enteros
df_merged[cols] = df_merged[cols].replace({'': 0}).fillna(0).astype(int)
# En columna date:
# Renombro columna por call_date para mayor claridad
df_merged.rename(columns={'date': 'call_date'}, inplace=True)
# Convertir la columna 'call_date' a tipo datetime para facilitar análisis posteriores
# para ello he utilizado la librería locale y una función de pandas, que he buscado en ChatGPT
import locale
locale.setlocale(locale.LC_TIME, 'Spanish_Spain')  # Windows (alternativa)
df_merged['call_date'] = pd.to_datetime(df_merged['call_date'], format='%d-%B-%Y', errors='coerce')


# Verificar resultado
df_merged.info()
#Hecho un vistazo a la cantidad de valores nulos
df_merged.isnull().sum()
```
