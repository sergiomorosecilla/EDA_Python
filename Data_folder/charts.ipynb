```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

```python
#Cargo el archivo csv
df_charts = pd.read_csv('./clean&merge_data.csv')

df_charts.head()
```

```python
# Defino las columnas categóricas que nos ineteresan
# para los gráficos que mostraremos

col_num = ['age', 'income']
```

```python
#Utilizamos un bucle for, para que nos haga todos los gráficos. Con 
#el bucle siempre podremos escalarlo a mas variables falcilmente añadiendo
#en variable creada col_num
for columna in col_num:
    plt.figure(figsize=(12, 5))

    sns.histplot(data=df_charts, x=columna, bins=20, kde=False)

    plt.title(f'Frecuencia de {columna}')
    plt.xlabel('')
    plt.ylabel('')
    plt.xticks(rotation=45)

    plt.show()
```

```python
# Defino las columnas categóricas que nos ineteresan
# para los gráficos que mostraremos
col_cat = ['job', 'marital', 'education', 'contact', 'poutcome',
       'subscribe']
```

```python
for columna in col_cat:
    plt.figure(figsize=(12, 5))

    sns.histplot(data=df_charts, x=columna, bins=20, kde=False)


    plt.title(f'Frecuencia de {columna}')
    plt.xlabel('')
    plt.ylabel('')
    plt.xticks(rotation=45)

    plt.show()
```

## Relaciones cruzadas

Vamos a seleccionar una columna, en este caso va a ser `subscribe`versus columnas numéricas para ver la correclación


```python
columna = 'income'

cols_num = df_charts.select_dtypes(include='number').columns

# Generar gráficos de dispersión
for col in cols_num:

    plt.figure()
    sns.scatterplot(x=df_charts[col], y=df_charts[columna], hue=df_charts['subscribe'], alpha=0.5)

    plt.show()

    
```
