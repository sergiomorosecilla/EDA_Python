```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

```python
#Cargo el archivo csv
df_charts = pd.read_csv('./clean&merge_data.csv')

df_charts.head()
```

```python
# Defino las columnas númericas que nos ineteresan
# para los gráficos de frecuencia que mostraremos

col_num = df_charts.select_dtypes(include='number').columns
col_num
```

```python
#Utilizamos un bucle for, para que nos haga todos los gráficos. Con 
#el bucle siempre podremos escalarlo a mas variables falcilmente añadiendo
#en variable creada col_num
for columna in col_num:
    plt.figure(figsize=(12, 5))

    sns.histplot(data=df_charts, x=columna, bins=20,kde=False)

    plt.title(f'Frecuencia de {columna}')
    plt.xlabel('')
    plt.ylabel('')
    plt.xticks(rotation=45)

    plt.show()
```

```python
# Defino ahora las columnas categóricas para las visualizaciones de 
#frecuencias
col_cat = ['job', 'marital', 'education', 'contact', 'poutcome',
       'subscribe']
```

```python
for columna in col_cat:
    plt.figure(figsize=(12, 5))

    sns.histplot(data=df_charts, x=columna, bins=20, kde=False)


    plt.title(f'Frecuencia de {columna}')
    plt.xlabel('')
    plt.ylabel('')
    plt.xticks(rotation=45)

    plt.show()
```

## Relaciones cruzadas

Vamos a seleccionar una columna, en este caso va a ser `income`versus columnas numéricas para ver la correclación y el grado de `subscribe`
(contratatación)

```python
columna = 'income'

cols_num = df_charts.select_dtypes(include='number').columns
col_num = cols_num.drop(['income',])

# Generar gráficos de dispersión
for col in cols_num:

    plt.figure()
    sns.scatterplot(x=df_charts[col], y=df_charts[columna], hue=df_charts['subscribe'], alpha=0.5)

    plt.show()

    
```

Revisamos ahora con la variables categóricas

```python
columna = 'income'

cols_cat = df_charts.select_dtypes(include='object').columns
#Elimino las que no deseamos
cols_cat = cols_cat.drop(['subscribe', 'call_date', 'nr.employed', 'id'])
cols_cat

```

```python
# Generar gráficos de dispersión
for col in cols_cat:

    plt.figure()
    sns.scatterplot(x=df_charts[col], y=df_charts[columna], hue=df_charts['subscribe'], alpha=0.5)
    plt.xticks(rotation=90)

    plt.show()
```

```python
# Definir la variable objetivo
columna = 'subscribe'

for col in df_charts.columns:
    # Saltar la variable objetivo
    if col == columna:
        continue

    plt.figure(figsize=(10, 5))

    if col in col_num:
        sns.histplot(x=df_charts[col], hue=df_charts[columna], palette="", legend=True)
    elif col in col_cat:
        sns.countplot(x=df_charts[col], hue=df_charts[columna], palette="magma")

    plt.title(f"{col} vs {columna}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.tight_layout ()
    plt.show()
```
